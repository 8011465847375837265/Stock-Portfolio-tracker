```python
print("Welcome to Stock Portfolio Tracker !")
print("Let's calculate your investment.")


stock_name=input("Enter stock name:").upper()

quantity=int(input("Enter quantity:"))

stock_prices = {
    "AAPL":180,
    "TSAL":250,
    "MSFT":420
}
price = stock_prices[stock_name]

total = price * quantity

print("you selected:",stock_name)
print(" Quantity:",quantity)
print("Stock price:",price)
print("Total investment:", total)
```

```python
import csv
```

```python
with open("stock_portfolio.csv","w",newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Stock Name","Quantity","Price","Total"])
    writer.writerow([stock_name,quantity,price,total])
    print("Stock details saved successfully!")
```
